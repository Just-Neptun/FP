```python
import numpy as np

# meters
l = 0.02
b = 5e-3
d = 200e-6
g_a = 100e-6
g_d = 100e-6

D_a = 3e-3
D_d = D_a


# Volts
U_0 = 10
U_a = U_0
U_B = 108

# N / m^2
E = 98e9

# Farad
C_L = 300e-12

# Vaccum Permittivity
epsilon_0 = 8.8541878188e-12

omega = 1

# ----- ----- ----- -----

S = np.pi * (D_a/2)**2

C_a = epsilon_0 * S / g_a
C_d = epsilon_0 * S / g_d

def force_on_end(t):
    return 1/4 * C_a * U_0**2 / g_a * (1 + np.cos(2 * omega * t))

def deflection_of_end(F):
    return 4 * l**3 / (E * d**3 * b) * F

def detector_potential(z):
    return U_B * z * C_d / (C_L * g_d)

max_U = detector_potential(deflection_of_end(force_on_end(0)))
print(max_U)

# print(2 * epsilon_0**2 * U_B * U_0**2 * l**3 / (E * d**3 * b) * S**2 / (g_d**2 * g_a**2) * 1/C_L)
```
